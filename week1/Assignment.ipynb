```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
# loading dataset and basic stats
df = pd.read_csv("student_data.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# handling missing data
df.isnull().sum()
```

```python
# filling age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

#  filling cgpa with median
df["CGPA"] = df["CGPA"].fillna(df["CGPA"].median())

# filling attendance with mean
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
```

```python
df.isnull().sum()
```

```python
#  encoding categorical variables
#  applying label encoding to gender
le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
```

```python
# applying one hot encoding to departrment
ohe = OneHotEncoder(sparse_output = False)
encoded = ohe.fit_transform(df[["Department"]])
encoded_df = pd.DataFrame(
    encoded,
    columns = ohe.get_feature_names_out(["Department"])
)
encoded_df
```
